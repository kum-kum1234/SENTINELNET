```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_curve, auc, confusion_matrix, classification_report
)
import seaborn as sns

```

```python
df = pd.read_csv("fridayworkinghrs.csv")
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.fillna(0, inplace=True)

print(" Columns in dataset:\n", df.columns.tolist())
```

```python
label_col = " Label" 
df[label_col] = df[label_col].astype(str)  
df[label_col] = df[label_col].apply(
    lambda x: 0 if x.strip().upper() == "BENIGN" else 1
)

print("Binary conversion complete!")
print(df[label_col].value_counts())
print("\n(0 = Normal, 1 = Attack)")
```

```python
counts = df[label_col].value_counts()

plt.figure()
plt.bar(counts.index, counts)
plt.xticks([0,1], ["Normal","Attack"])
plt.xlabel("Traffic Type")
plt.ylabel("Count")
plt.title("Network Traffic: Attack vs Normal")
plt.show()

plt.figure()
plt.pie(counts, labels=["Normal","Attack"], autopct="%1.1f%%")
plt.title("Attack vs Normal %")
plt.show()
```

```python
X = df.drop(columns=[label_col])
y = df[label_col]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print("Train size:", len(X_train), "Test size:", len(X_test))
```

```python
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1, class_weight="balanced"),
    "Extra Trees": ExtraTreesClassifier(n_estimators=200, random_state=42, n_jobs=-1, class_weight="balanced"),
    "Naive Bayes": GaussianNB(),
    "XGBoost": XGBClassifier(n_estimators=200, random_state=42, tree_method="hist"),
    "LightGBM": LGBMClassifier(n_estimators=200, random_state=42, device="cpu"),
    "Logistic Regression": LogisticRegression(max_iter=1500, C=5, solver="lbfgs", random_state=42, n_jobs=-1, class_weight="balanced")
}

for name, model in models.items():
    print(f"Training {name}...")
    model.fit(X_train, y_train)
    print("Done")
```

```python
results = []
from sklearn.metrics import roc_curve, auc, accuracy_score, precision_score, recall_score, f1_score, f1_score

for name, model in models.items():
    y_train_pred = model.predict(X_train)
    y_test_pred  = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
    else:
        y_proba = model.decision_function(X_test)
        y_proba = (y_proba - y_proba.min())/(y_proba.max()-y_proba.min())

    fpr, tpr, _ = roc_curve(y_test, y_proba)
    auc_score = auc(fpr, tpr)

    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc  = accuracy_score(y_test, y_test_pred)

    results.append([
        name,
        train_acc,
        test_acc,
        precision_score(y_test, y_test_pred, zero_division=0),
        recall_score(y_test, y_test_pred, zero_division=0),
        f1_score(y_test, y_test_pred, zero_division=0),
        auc_score
    ])

metrics_df = pd.DataFrame(results, columns=[
    "Model", "Train Acc", "Test Acc", "Precision", "Recall", "F1", "AUC"
])

print("\nMODEL PERFORMANCE\n")
print(metrics_df.to_string(index=True))
```

```python
best_model_name = metrics_df.loc[metrics_df["Test Acc"].idxmax(), "Model"]
best = models[best_model_name]

y_best_pred = best.predict(X_test)
cm = confusion_matrix(y_test, y_best_pred)

print("Best Model Selected:", best_model_name)

plt.figure()
sns.heatmap(cm, annot=True, fmt="d", cmap=None)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
if hasattr(best, "predict_proba"):
    y_proba = best.predict_proba(X_test)[:, 1]
else:
    y_proba = best.decision_function(X_test)
    y_proba = (y_proba - y_proba.min())/(y_proba.max() - y_proba.min())

fpr, tpr, _ = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.4f}")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC AUC Curve")
plt.legend()
plt.show()

print("AUC Score:", round(roc_auc, 5))
```

```python
print("SUMMARY\n")
print("Dataset shape:", df.shape)
print("Binary classification:", {0: "Normal", 1: "Attack"})
print("Scaler used:", scaler.__class__.__name__)
print("Train samples:", len(X_train))
print("Test samples :", len(X_test))
print("Best Model:", best_model_name)
```
